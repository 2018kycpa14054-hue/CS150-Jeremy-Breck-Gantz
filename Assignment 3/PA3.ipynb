# Task 2: Student Grades And At-Risk Status 

# A. Read Five Input Grade Averages (Which Should Be Floats) Representing Student Grades At The Mid-Term Break Into An Initially Empty List Called student_grades.

```python
student_grades=[]
student_grade1=float(input("Enter First Student Grade:"))
student_grades.append(student_grade1)
student_grade2=float(input("Enter Second Student Grade:"))
student_grades.append(student_grade2)
student_grade3=float(input("Enter Third Student Grade:"))
student_grades.append(student_grade3)
student_grade4=float(input("Enter Fourth Student Grade:"))
student_grades.append(student_grade4)
student_grade5=float(input("Enter Fifth Student Grade:"))
student_grades.append(student_grade5)
```

# B. First Calculate And Print The Class Average, Minimum Grade, And Maximum Grade.

```python
class_average=sum(student_grades)/len(student_grades)
minimum_grade=min(student_grades)
maximum_grade=max(student_grades)
print("The Class Average Is", class_average)
print("The Class Minimum Is", minimum_grade)
print("The Class Maximum Is", maximum_grade)
```

# C. Any Student With A Grade Below 65 Is Considered “At Risk” – As Such, You Should Print A Message Indicating How Many Of The Students In The Course Are At Risk.  

```python
at_risk=0

for grade in student_grades:
    if grade <65: 
        at_risk= at_risk+1
    
print("The Number Of Students In The Course That Are At-Risk Is", at_risk)
```

# Task 3: Ranges And Multiples 

# A. Create A List Called mult_of_2 That Contains All Multiples Of 2 Between 1 And 100

```python
mult_of_2=list(range(2,101,2))
print(list(mult_of_2))
```

# B. Create A List Called mult_of_3 That Contains All Multiples Of 3 Between 1 And 100.

```python
mult_of_3=list(range(3,101,3))
print(list(mult_of_3))
```

# C. Create A List Called mult_of_both That Contains All Elements That Are Present In Both mult_of_2 and mult_of_3.

```python
mult_of_both=[]
for number in mult_of_2: 
        if number in mult_of_3: 
            mult_of_both.append(number)
print(mult_of_both)
```

# D. Create A Fourth List Called mult_of_either That Contains All Numbers Present In Either mult_of_2 or mult_of_3

```python
mult_of_either=[]
for number in mult_of_2: 
        if number not in mult_of_either: 
            mult_of_either.append(number)
for number in mult_of_3: 
        if number not in mult_of_either: 
            mult_of_either.append(number)
print(mult_of_either)
```
