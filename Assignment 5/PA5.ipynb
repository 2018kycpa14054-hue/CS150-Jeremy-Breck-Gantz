## Task 1: Weekend Activities As Sets 
## A. Let a user input a series of activities for a Saturday, and store these in a set called actSat.  The word “Done” should stop the process (no more activities).   Remember, this is a set, so if a user inputs an activity more than once, only one will be stored.
## B. Repeat the process to let the user input a series of activities for Sunday, and store these in a set called actSun.
## Sample input for steps A and B are as follows:
## Enter an activity from last Saturday (“Done” to stop):
## Eating
## Enter an activity from last Saturday (“Done” to stop):
## Working
## Enter an activity from last Saturday (“Done” to stop):
## Eating
## Enter an activity from last Saturday (“Done” to stop):
## Jogging
## Enter an activity from last Saturday (“Done” to stop):
## Done
## Enter an activity from last Sunday (“Done” to stop):0
## Eating
## …
## C. Print both sets 
## D. Print the series of activities done on both days (Saturday and Sunday)
## E. Print the series of activities done on at least one of the days (Saturday or Sunday)


```python
# A. Read Saturday Activities From The User Into A Set
actSat=set()
activity=input("Please Enter An Activity You Did On Last Saturday ('Press Done' to stop:\n")
while activity !="Done": 
    actSat.add(activity)
    activity=input('Enter An Activity From Last Saturday ("Done" to stop):\n')
# B. Read Sunday Activities From The User Into A Set
actSun=set()
activity=input("Please Enter An Activity You Did On Last Sunday ('Press Done' to stop:\n")
while activity !="Done": 
    actSun.add(activity)
    activity=input('Enter An Activity From Last Sunday ("Done" to stop):\n')

# C. Print Both Days' Activity Sets
print("\nSaturday Activities:",actSat)
print("\nSunday Activities:", actSun)

# D. Print Activities Done On Both Days
print("Activities Done On Both Days:",actSat&actSun)

# E. Print Activities Done On At Least One Of The Days
print("Activities Done On At Least One Of The Days:", actSat|actSun)
```

## Task 2: Triangle Calculations 
## Write a function triangle_solve(leg1, leg2) that takes as input the length of two legs of a right triangle.  Based on this information, you should be able to compute the triangle’s hypotenuse length, perimeter, and area.  These should be returned as a three-element tuple (hypotenuse, perimeter, area) from the function. 
## Ex: if triangle_solve(3, 4) is called, the tuple (5, 12, 6) will be returned.


```python
import math
def triangle_solve(leg1, leg2): 
    hypotenuse=math.sqrt(leg1**2 + leg2**2)
    perimeter= leg1 + leg2 + hypotenuse
    area=(leg1*leg2)/2
    return(hypotenuse, perimeter, area)
# Print Out The Results
print(triangle_solve(3,4))
```

## Task 3: Calculator Class
## Create a class named Calculator with the following methods: 
## • add(a, b): Perform addition and returns the result.
## • subtract(a, b): Perform subtraction and returns the result. 
## • multiply(a, b): Perform multiplication and returns the result. 
## • divide(a, b): Perform division and returns the result. 
## • power(a, b): Raises the first  quantity (a) to the second (b) and returns the result.
## In addition, Calculator should maintain the most recent calculation result as an attribute called prev_result, as well as an attribute total_calculations which maintains the total number of calculations performed (any call to add, subtract, multiply, divide, or power counts.)
## Lastly, Calculator should have an additional method called info(), which should print out the current value for prev_result and total_calculations.
## Note: while you technically don’t have to test your function for task 2 and class for task 3, it is highly recommended that you do so.


```python
class Calculator: 
    def __init__(self): 
        self.prev_result=None
        self.total_calculations=0
    def add(self,a,b): 
            result=a+b
            self.prev_result=result
            self.total_calculations+=1
            return result
    def subtract(self,a,b): 
            result=a-b
            self.prev_result=result
            self.total_calculations+=1
            return result
    def multiply(self,a,b): 
            result=a*b
            self.prev_result=result
            self.total_calculations+=1
            return result
    def divide(self,a,b): 
            result=a/b
            self.prev_result=result
            self.total_calculations+=1
            return result

    def power(self,a,b): 
            result=a**b
            self.prev_result=result
            self.total_calculations+=1
            return result
    def info(self): 
            print("Previous Result:",self.prev_result)
            print("Total Calculations:",self.total_calculations)

# Print It 
calc=Calculator()
print(calc.add(3,4))
print(calc.subtract(10,6))
print(calc.multiply(7,2))
print(calc.divide(9,3,))
print(calc.power(2,3))
calc.info()
```
